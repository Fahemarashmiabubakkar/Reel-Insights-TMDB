```python
import pandas as pd
from sqlalchemy import create_engine, text
```

```python
engine = create_engine(
    "mysql+pymysql://root:root@localhost:3306/TMDB"
)

print("SQLAlchemy connection created successfully!")
```

```python
with engine.connect() as connection:
    result = connection.execute(text("SELECT DATABASE()"))
    print(result.fetchone())
```

```python
from sqlalchemy import text

with engine.begin() as connection:

    connection.execute(text("""
        CREATE TABLE IF NOT EXISTS movies (
            movie_id INT PRIMARY KEY,
            title VARCHAR(255),
            original_language VARCHAR(10),
            release_date DATE,
            runtime INT,
            budget BIGINT,
            revenue BIGINT,
            popularity FLOAT,
            vote_average FLOAT,
            vote_count INT,
            status VARCHAR(50),
            overview TEXT
        )
    """))

print("Movies table created successfully!")
```

```python
movies = pd.read_csv(
    r"c:\xampp\perl\vendor\lib\auto\Devel\vscode\TMDB_data\cleaned_movies.csv"
)

print("Movies dataset loaded successfully!")
```

```python
print(movies.shape)
print(movies.columns.tolist())
```

```python
movies_sql = movies[
    [
        "movie_id",
        "title",
        "original_language",
        "release_date",
        "runtime",
        "budget",
        "revenue",
        "popularity",
        "vote_average",
        "vote_count",
        "status",
        "overview"
    ]
].copy()
```

```python
movies_sql.to_sql(
    name="movies",
    con=engine,
    if_exists="append",
    index=False,
    chunksize=1000
)

print("Movies data inserted successfully!")
```

```python
with engine.begin() as connection:

    connection.execute(text("""
        CREATE TABLE IF NOT EXISTS movie_genres (
            movie_id INT,
            genre_id INT,

            PRIMARY KEY (movie_id, genre_id),

            FOREIGN KEY (movie_id)
                REFERENCES movies(movie_id),

            FOREIGN KEY (genre_id)
                REFERENCES genres(genre_id)
        )
    """))

print("Movie genres table created successfully!")
```

```python
movie_genres.to_sql(
    name="movie_genres",
    con=engine,
    if_exists="append",
    index=False,
    chunksize=1000
)

print("Movie genres data inserted successfully!")
```
